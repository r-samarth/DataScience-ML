## OOPS
Object oriented programming 


```python
class dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def brak(self):
        print(f"{self.name} says boooh")
    def age(self):
        return self.age
    
dog1=dog('tommy',18)
dog1.brak()
print(dog1.age)
```

```python
class Bank:
    def __init__(self,owner,age,amount):
        self.owner=owner
        self.age=age
        self.amount=amount
        print(f"Account Created for {self.owner} who is {self.age} years  old with the Account balance {self.amount} \n ---Thnak YOU!!---")
    def deposit(self,amount):
        self.amount+=amount
        print(f"{amount} deposited to the account,owner name {self.owner}")
    def withdraw(self,amount):
        self.amount-=amount
        print(f"{amount} withdrawaed from the account, owner name {self.owner}")
    def showbalance(self):
        print(f"Current Balance in {self.owner}'s account is {self.amount}")

acc1=Bank('Sam',18,400)
acc1.deposit(400)
acc1.showbalance()
acc1.withdraw(700)
acc1.showbalance()

```

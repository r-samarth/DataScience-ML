## File operancetions in python 


```python
## read a whole file

with open ('example.txt',mode='r',newline='') as file :
    print(file.read())
```

```python
## reading line by line 
with open ('example.txt',mode='r',newline='') as file :
    for i in file :
        print(i)
```

```python
# writeing in a file (over writhing)
with open ('example.txt', mode='w', newline='') as file :
    file.write(input(""))
```

```python
## without ovetr writing 
with open('example.txt',mode='a',newline='') as file :
    file.write(' \n sam here ')
```

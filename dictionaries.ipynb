### introduction to Dictionaries
 
Dictionaries are unoredared collection of iteams. they store data in key-value pairs key must be unike and immuatable 


```python
# creating dictionaries 
dic1={}
print(type(dic1))
```

```python
dic1=dict()
print(dic1)
```

```python
dic1={"Name":"Sam","Sem":"1st","sec":"CSE-G"}
print(dic1,type(dic1))
```

```python
## Only one key can be used in dict 
dic2={"Name":"Sam","Sem":"1st ","sec":"CSE-G","Name":"Su"}
print(dic2)## the name will be replased with the new one 
```

```python
## accssing elements 
print(dic1['Name'])
print(dic1['Sem'])

## accssing elements useing get
print(dic1.get('Name'))
print(dic1.get('last_name')) ##  when the key is not present in the dic it returns "none"
print(dic1.get('last_name','not available'))## giveing a default value to the dic to reture if itz none

```

```python
## modifying  a dic
## since the dics ate mutable we can use add,update and delete 
dic1={"Name":"Sam","Sem":"1st","sec":"CSE-G"}
dic1['Sem']='2nd'
print(dic1)
dic1['age']=18
print(dic1)
```

```python
del dic1['age'] # deleting a key with the value 
print(dic1)
```

```python
## methodes on dict
keys=dic1.keys()
vlaues=dic1.values()
item=dic1.items()
print(keys)
print(vlaues)
print(item)
```

```python
## normal copy 

dic3=dic1
dic1['Name']='soooo'
print(dic1)
print(dic3)
## both even in copyed dict will be see the chnages its a big problem
```

```python
##shallow copy (this will creat a seprate memory for the copyed dict)
dic2=dic1.copy()
dic1['Name']='sam'
print(dic1)
print(dic2)
```

```python
##itreating over dict

## itereating over keys 
print("keys :  ")
for i in dic1.keys():
    print(i)
print("vlaues : ")
## itereating over values
for j in dic1.values():
    print(j)
print("items : ")
#itereating over iteams
for k in dic1.items():
    print(k)
#itereating over iteams(pirs)
for x,y in dic1.items():
    print(f"{x}:{y}")
```

```python
## nestered dict's
students={
    "srtd 1":{"name":"saman","age":18},
    "srtd 2":{"name":"superman","age":99},
    "srtd 3":{"name":"batman","age":89}
}
print(students)
```

```python
print(students["srtd 1"]["name"])
print(students["srtd 2"]["name"])
print(students["srtd 3"]["name"])
```

```python
print(students.items())
```

```python
##itrating over nested dict
for i,j in students.items():
    print(f"{i}:{j}")
    for k,l in j.items():
        print(f"{k}:{l}")
    print()
```

```python
## dict compretion 
seq={x:x**2 for x in range(6)}
print(seq)
```

```python
seqe={x:x**2 for x in range(6) if x%2==0}
print(seqe)
```

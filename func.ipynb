```python
##positional arguments
def fun(*args):
    for i in args:
        print(i)
```

```python
fun(2,3,2.4,"sam",True)
type(fun())
```

```python
## keywords arguments 
def sam(**kwargs):
    for i,j in kwargs.items():## this works as a dict
        print(f"{i}:{j}")
```

```python
sam(name="samarth",age=18,sem="2nd")
```

```python
## both the arguments together
def sam2(*args,**kwargs):
    for i in args:
        print(f"PA's = {i}")        ## PA = positional arguments
    print("KA's :")                 ## KA = keywords arguments
    for j,k in kwargs.items():
        print(f"{j}:{k}")
```

```python
sam2(1,2,3,2,1,4,5,3,name="samarth",age=18)
```

# Examoles of functions 

1. Temperature converter 

```python
def temp(tem,unit):
    if unit == 'C':
        return tem * 9/5 + 32 
    elif unit == 'F':
        return (tem-32)*5/9
    else :
        return None
        
    
    
```

```python
temp(71,'F')
```

```python
temp(21,'C')
```

2. strong password checker 

```python
def pas(p):
    if (len(p))<8:
        return False
    if not any(i.isdigit() for i in p):
        return False
    if not any(i.islower() for i in p):
        return False
    if not any(i.isupper() for i in p):
        return False
    if not any(i in '!@#$%&*()_+' for i in p):
        return False
    return True
```

```python
print(pas("Sam@2007"))
```

```python
print(pas("Sam2007"))
```

4. finding the factroal of the number 

```python
def fact(n):
    if n==1:
        return 1
    else:
        return n*(fact(n-1))
        
```

```python
print(fact(int(input("Enter a number"))))
```

## Lambda functions 

its just a anouther type of function in which there is no function name and it is alos called as Anonymous function 

```python
def add(x,y):
    return x+y
print(add(4,4)) ## normal function 

add1=lambda a,b:a+b
print(add1(1,2)) #lambda function 
```

```python
even= lambda num:num%2==0
even(8)
```

```python
L=[2,4,1,5,6,7,3]
list(map(lambda x:x*x,L)) ## map function in py 
```

## map function 

A map function apples a given function to all inputs in a given list(or any other itrable) and returns a map object(an iterater)

```python
## cube of a number
def cub(x):
    return x**3

a=[1,3,4,5,6,7,8,9]
list(map(cub,a))
```

```python
a=[1,3,4,5,6,7,8,9]
list(map(lambda x:x**3,a)) ## useing a lambda function 
```

```python
## adding two numbers in a list and displaying that in the from of another list 
l1=[4,5,6]
l2=[1,2,3]
list(map(lambda a,b:a+b,l1,l2))
```

```python
a1=['2','3','4']
list(map(lambda x:int(x),a1)) ## useing map+lambda 
```

```python
a1=['2','3','4']
list(map(int,a1)) ## useig only map
```

## filter function 

It is used to filter out the items form a list based on a conduction 


```python
x=[1,2,3,4,5,6,11,22,32,12,44,56,43]
print(x)
```

```python
list(filter(lambda x:x%2==0,x))
```

```python
list(filter(lambda x:x>10 and x%2==0,x))
```

```python
ppl=[
    {'name':'sam','age':18},
    {'name':'john','age':12},
    {'name':'joy','age':19}
]
```

```python
list(filter(lambda x:x['age']>18,ppl))
```

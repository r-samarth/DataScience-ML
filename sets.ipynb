```python
## We can have even lists and tuples in the set 
lis=[i for i in range(11)]
print(f"list = {lis}")
my_set2=set(lis)
print(f"list convearted set = {my_set2}")

tu=tuple((j for j in range(21)))
print(f" Tuple = {tu}")
my_set2=set(tu)
print(f"tuple  convearted set = {my_set2}")
```

```python
## adding a element to the set  
print(my_set)
my_set.add(9)
print(my_set)
```

```python
## removeing an element form the set
my_set.remove(2)
print(my_set)
```

The probleam in remove function  is that it shows a error if the element is not present in the set
to fix it we use the function called discard it removes the element from the set and does not show an erroe if the element is not present in the set 


```python
## error from remove 
my_set.remove(21)
print(my_set)
```

```python
## discard function 
my_set.discard(21)
print(my_set)
```

```python
## pop returns the first element and removes it fron the set 
ele=my_set.pop()
print(ele)
print(my_set)
```

```python
## clear all the elements 
my_set.clear()
print(my_set)
```

# mathematical operations on set

```python
## mathematical operations on set 
set1={1,2,3,4,5,6}
set2={5,6,7,8,9,10}
print(f"set 1 ={set1}")
print(f"set 2 ={set2}")

##Union
uset=set1.union(set2)
print(f"The union set = {uset}")

## intersection 
iset=set1.intersection(set2)
print(f"The intersection set = {iset}")


```

After unien and intersetin we have updated intersection which finds the intersecion of the sets and updates in the given set 


```python
set1.intersection_update(set2)
print(set1)
```

```python
set1={1,2,3,4,5,6}
## differance 
print(set1.difference(set2))
print(set2.difference(set1))


```

```python
#differnce update 
set1.difference_update(set2)
print(set1)
```

```python
##symmetric difference (the intersecion of both the sets will be removed)
set1={1,2,3,4,5,6}
print(set1.symmetric_difference(set2))
```

```python
#methodes used in sets 
set1 ={1,2,3,4}
set2={2,3}
print(set2.issubset(set1))
print(set1.issubset(set2))
print(set1.issuperset(set2))
```

```python
## removing same same words in a list
li=[1,2,3,4,5,6,1,2,3,8,9,5,6,7,8]
s=set(li)
print(s)
```

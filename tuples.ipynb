All the operations/slicing  performed in lists can be done in tuple also 


```python
## lets see some examoles of operactios
li=[i for i in range(1,11)] 
tu=tuple(li)
print(tu[0])
print(tu[:5])
print(tu[:-5])
print(tu[3:6])
print(tu[::-2])




```

packing and unpacking a tuple 


```python
## packing 
mix=1,"hi","bro",'s','n',1.5,True
print(mix)# directly it will take it as tuple 
```

```python
#unpacking
a,b,c,d,e,f,g = mix
print(a)
print(b)
print(c)
print(g)
```

```python
#unpacking with *
ti=1,2,'e','r',4,"gggggg"
fi,*mid,lat=ti
print(fi)
print(mid)
print(lat)
```

nested tuples and lists 


```python
## nested list
liss=[["sam","nis",13,18],["su","so",20,'x'],["dig","me",1,'y']]
print(liss)
```

```python
#nested tuples 
tuss=(("sam","nis",13,18),("su","so",20,'x'),("dig","me",1,'y'))
print(tuss)
```

```python
print(tuss[1])
print(tuss[2][:2])
print(tuss[1:3])
```

```python
## tuple inside a list 
lisx=[["sam","nis",13,18],["su","so",20,'x'],("dig","me",1,'y')]
print(lisx)
print(lisx[1])
print(lisx[2][:2])
print(lisx[1:3])
```

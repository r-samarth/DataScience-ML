```python
#csv
import csv 

with open('example.csv',mode='w',newline='') as file :
    wri=csv.writer(file)
    wri.writerow(['Nmae','age'])
    wri.writerow(['sam','18'])
with open('example.csv',mode='r') as file:
    rir=csv.reader(file)
    for i in rir:
        print(i)
```
